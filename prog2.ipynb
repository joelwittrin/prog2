Type hinting:

Any Python expression has a type, i.e. 'int', 'list' etc. Some languages fix variables to specific types for performance. In Python this is done through type hinting, e.g. 'x: int'. 

This can also express what is to be returned from a function with an arrow '->', e.g. 'function(x: int) -> str:'. 
'z: float = function()' would then yield an error as the expected return type is string, not float.
Returning the same type would look like 'function(x: int) -> x'.

The concept is more helpful for complex types, e.g.:
'def get_names_to_age() -> dict[str, int]:' then returns a dictionary with strings and integers. 

An expression with interchangeable types is called a Union, e.g. 'reverse(whhat: str | list):'.

The point with type hinting is to help with debugging and prohibit the code being used wrong. Note that it guarantees the type, NOT the value of said type. 


Similar to type hint, asserts are used to require certain values or types from an argument. Asserts are a way to signal intent to the reader:

```python
def factorial(val: int):
    assert val >= 0, 'Value should be non-negative!'
    ...
```

**Complexity analysis**

Performance can have two main dimensions: *What amount* of work and information is required and *what is the efficency* of the implementation. As the problem increases, the asymptotic behaviour is measured. 

Consider a search function that returns a searched value in a list, i.e. 'search(x, lst)' returns x from lst, with a help function that implements binary search (we assume that the list is sorted). We split the list for each iteration, such that

n/2 -> n/4 -> ... -> n/(2^k), meaning that the problem decreases for each iteration. For many iterations, the asymptotic behaviour is 1/(2^k), i.e. O(2^k).

Consider a list with n being used with 'search(x, lst)'. The complexities for different scenarios are:

Best case: Theta(1)
Worst case: Theta(n)
Average: Theta(n), since n/2 goes to n for large n, hence we ignore the constant 1/2

Theta(nlogn) << Theta(n^2) in time, i.e. problems with the first complexity is much faster for large n. Remember that this is a trade-off depending on what algorithm is used, and that Theta is rarely exact due to different computer power.

A *datastructure* is a way of organizing a set of data, e.g. a registry for cars. The objects within the datastructure are called *posts*. A datastructure can be modified through operations such as editing, deleting, search etc. Common datastructures are arrays, trees, linked lists and hash tables.

Arrays: storage effective and easy to search with a complexity of Theta(1), which turns into Theta(n) when adding a post in the structure as all following items needs to be moved, e.g:

lst_before = np.array([5, 8, 13, 34, 55]) --> lst_adding21 = np.array([5, 8, 13, 21, 34, 55])

Here, adding 21 requires 34 and 55 to move one spot. This complexity is solved through *linked lists*, where posts "point" to one another in different memories through an *adress*. This results in stored items not being moved but a search complexity of Theta(n).
```python
# A linked list is initiated through the following structure

class Node:
    def __init__(self, data, succ):
        self.data = data
        self.succ = succ

# Create the actual list, starting from the back. Write first = Node(value, None) for the last node, while all others are written as first = Node(value, first), i.e.

first = Node(17, None)
first = Node(5, first)
first = Node(3, first)

sum = 0
while first:
    sum += first.data
    first = first.succ
print('Sum:', sum)
```

If we want to edit this linked list, we section through the different cases: adding to an empty list, the start of the list, the end of the list or the middle of the list. Assuming *x* is the value being added, the algorithm looks like:

```python
# Adding to an empty list or the first index
if first == None or x <= first.data:
    first = Node(x, first)

# Adding to the end or in the middle
else:
    while first.succ != None and x > first.succ.data:
        first = first.succ
    first.succ = Node(x, first.succ)
```

To avoid having to specify where and how *x* is to be inserted, the class Node() can be nestled within another class, such that only simple commands are used. This also avoids breaking the list if *first* is incorrectly pointed. Utilizing the algorithm above specifying how to insert *x*, setting the list to empty, the code becomes:

```python
class LinkedList:

    class Node:
        def __init__(self, data, succ):
            self.data = data
            self.succ = succ

    def __init__(self):
        self.first = None

    def insert(self, x):
        if first == None or x <= first.data:
            first = Node(x, first)
        else:
            while first.succ != None and x > first.succ.data:
                first = first.succ
            first.succ = Node(x, first.succ)

    # Print the list for pedagogical reasons
    def print(self):
        print('(', end = '')
        f = self.first
        while f:
            print(f.data, end='')
            f = f.succ
            if f:
                print(',', end = '')
        print(')')
```

The main takeaway from this block of code is that 'f = self.first' yields the index for start of the list. Implementing for or while loops become easier, as each value in the list can be iterated through by 'f = f.succ' which is the same as writing 'f = self.first.succ' (both within the class LinkedList). The logic comes naturally after trying to implement the code. Note that f is a variable pointing to self.first and does NOT overwrite it; if the first node is to be changed, one must explicitly write 'self.first = ...'!

One can also step through each node by continuing the .succ command, i.e. f.succ.succ.succ would step into a node three iterations away. This is useful when removing a node, as the current node no longer can point to it. Consider the list

5 --> 8 --> 13 --> 21 --> 34 --> 55

If 8 is removed, 5 must point to 13 for the chain not to break and thus, the .succ needs to be updated to .succ.succ. The core algorithm will then look something like

```python
f = self.first
while f.succ:
    if f.succ.data == x:
        f.succ = f.succ.succ
```

This chapter focuses on creating linked lists through classes, meaning that we want to retain the same functionality as a normal list. Commands such as 'for i in range' does not work, since Python has no built in function for iterating through a class. Thus, building an iteration function within the class is very helpful. Considering the established class, a function that solves this would be similar to:

```python
def __iter__(self):
    current = self.first
    while current:
        yield current.data
        current = current.succ
```

The command 'yield' implements a generator in Python automatically, allowing the memory to store the current iteration value. Instead of returning a value in a function, completely stopping the algorithm, yielding pauses the function and lets the code access the current stage when it is called upon again. The algorithm above simply pauses the function, until it is called again and 'current = current.succ' is continued. This function allows for the command 'for x in self' when established in a class. More specifically for linked lists, it iterates through the entire list without having it explicitly written out.

Similar to the 'iter'-function allowing a for loop implemented into the class, 'if x in list' can also be obtained through another magic function:

```python
def __contains__(self, x):
    for d in self:
        if d == x:
            return True
        elif x < d:
            return False
    return False
```

The point of the double-underscore functions are that Python connects the syntax without requiring the function being called upon, e.g.

if lst.contains_value(x): --> if x in lst:

which leads to a more readable code and shorter syntax. The important part of the 'contains'-function is 'for d in self', as this iterates through the list. Note that self can be any object and _ _ iter _ _ is not bound to a certain class!

Monte-Carlo:

Consider a circle within a square, such that r = 1 and the square 2x2 units. In order to approximate pi, one can obtain the formula pi = 4 * A_circ / A_squa. In order to simulate Monte Carlo, let (x,y) = [-1, 1] x [-1, 1] in a uniform random distribution, leading to n amount of points placed in the region. Separate into n_in and n_out relative to the circle within the square. Higher n leads to a better approximation and the formula can then be approximated as pi = 4 * n_in / n, where n = n_in + n_out.
